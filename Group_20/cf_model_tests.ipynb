## Imports

```python
import pandas as pd
from cf_model import CF_Model
from pathlib import Path
```

## Load data

```python
PATH = Path("./data/ebnerd_demo")
data_split = "train"
df_behaviors_train = pd.read_parquet(PATH.joinpath(data_split, "behaviors.parquet"))
df_history_train = pd.read_parquet(PATH.joinpath(data_split, "history.parquet"))
df_articles = pd.read_parquet(PATH.joinpath("articles.parquet"))
```

## Load model

```python
model = CF_Model(df_articles, df_behaviors_train, df_history_train)
```

```python
model.build_model()
```
